```python
import pandas as pd

path = r'C:\Users\AVINASH\Downloads\DataScience Assessment'

df = pd.read_csv(path + r'\airport_hourly.csv')
df['hour_ts'] = pd.to_datetime(df['hour_ts'])

# Check zone types first
print("Zone types:")
print(df['zone_type'].value_counts())

# Airport rows
airport = df[df['zone_type'] == 'airport_terminal'].copy()

# Hourly analysis
hourly = airport.groupby(airport['hour_ts'].dt.hour).agg(
    requests=('requests', 'sum'),
    fulfilled=('fulfilled_requests', 'sum'),
    unfulfilled=('unfulfilled_requests', 'sum'),
    captains=('online_captains', 'mean'),
    eta=('avg_eta_min', 'mean'),
    surge=('avg_surge_multiplier', 'mean')
).reset_index()

hourly['unfulfilled_%'] = (
    hourly['unfulfilled'] / hourly['requests'] * 100
).round(2)

hourly['fulfillment_%'] = (
    hourly['fulfilled'] / hourly['requests'] * 100
).round(2)

print("\nHourly airport performance:")
print(hourly.sort_values('unfulfilled_%', ascending=False).to_string(index=False))

# Overall
print("\nOverall airport:")
print(f"Requests: {airport['requests'].sum():,.0f}")
print(f"Unfulfilled: {airport['unfulfilled_requests'].sum():,.0f}")
print(
    f"Unfulfilled rate: "
    f"{airport['unfulfilled_requests'].sum() / airport['requests'].sum() * 100:.2f}%"
)
```
