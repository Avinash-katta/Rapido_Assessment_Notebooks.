```python
import pandas as pd

path = r'C:\Users\AVINASH\Downloads\DataScience Assessment'

h = pd.read_csv(path + r'\airport_hourly.csv')
t = pd.read_csv(path + r'\airport_trips.csv')

h = h[h['zone_type'] == 'airport_terminal'].copy()
h['hour'] = pd.to_datetime(h['hour_ts']).dt.hour

# Peak shortage window from B1
h['unfulfilled_%'] = h['unfulfilled_requests'] / h['requests'] * 100
peak = h[h['hour'].isin([21,22,23,0,1,2,3])]

print("B3: Peak airport shortage (21:00-03:00)")
print("Requests:", peak['requests'].sum())
print("Unfulfilled:", peak['unfulfilled_requests'].sum())
print("Unfulfilled %:", round(peak['unfulfilled_requests'].sum()/peak['requests'].sum()*100,2))

# Post-airport experience from B2
t['hour'] = pd.to_datetime(t['request_ts']).dt.hour
peak_t = t[t['hour'].isin([21,22,23,0,1,2,3])]

print("\nB3: Post-airport experience")
print("Return fare overall %:", round(t['got_return_fare_within_20min'].mean()*100,2))
print("Return fare peak hours %:", round(peak_t['got_return_fare_within_20min'].mean()*100,2))

print("\nReturn fare by drop zone:")
print((t.groupby('drop_zone_type')['got_return_fare_within_20min']
       .mean()*100).round(2))

# Simple intervention comparison
print("\nRecommendation:")
print("High shortage + low return-fare rate -> improve airport economics/return supply before relying only on acquisition.")
```
