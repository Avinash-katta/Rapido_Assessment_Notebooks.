```python
import pandas as pd
path = r'C:\Users\AVINASH\Downloads\DataScience Assessment'
airport_trips = pd.read_csv(path+r'\airport_trips.csv')
airport_hourly = pd.read_csv(path+r'\airport_hourly.csv')

print("Airport trips columns:", airport_trips.columns.tolist())
print(airport_trips.head())

# Let's inspect drop_zone_type, captain_cancelled, got_return_fare_within_20min
print("\nDrop zone types:")
print(airport_trips['drop_zone_type'].value_counts())

print("\nReturn fare within 20 min stats:")
print(airport_trips['got_return_fare_within_20min'].value_counts(normalize=True)*100)

print("\nCaptain cancelled stats:")
print(airport_trips['captain_cancelled'].value_counts(normalize=True)*100)
```

```python
# Let's analyze return fare probability by drop zone type and trip distance
print("Return fare within 20 min by drop zone type:")
print(airport_trips.groupby('drop_zone_type')['got_return_fare_within_20min'].mean() * 100)

print("\nAverage trip distance by drop zone type:")
print(airport_trips.groupby('drop_zone_type')['trip_distance_km'].mean())

print("\nReturn fare within 20 min by whether trip distance > median:")
median_dist = airport_trips['trip_distance_km'].median()
airport_trips['long_trip'] = airport_trips['trip_distance_km'] > median_dist
print(airport_trips.groupby('long_trip')['got_return_fare_within_20min'].mean() * 100)
```
